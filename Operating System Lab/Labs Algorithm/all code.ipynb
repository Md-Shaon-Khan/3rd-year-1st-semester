**Banker's Algorithm**

```python
n = int(input("Enter the number of processes:"))
m = int(input("Enter the number of resources:"))

allocation = []
maximum = []
available = []
need = []

print("Enter the allocation matrix: ")
for i in range(n):
    row = []
    for j in range(m):
        value = int(input(f"Enter P{i+1} R{j+1}"))
        row.append(value)
    allocation.append(row)

print("Enter the maximum matrix: ")
for i in range(n):
    row = []
    for j in range(m):
        value = int(input(f"Enter P{i+1} R{j+1}"))
        row.append(value)
    maximum.append(row)

print("Enter the available resources:")

for j in range(m):
    available.append(int(input(f"R{j+1}: ")))

for i in range(n):
    row = []
    for j in range(m):
        row.append(maximum[i][j] - allocation[i][j])
    need.append(row)

completed = []
while len(completed) < n:
    found = False

    for i in range(n):
        if i not in completed:
            possible = True

            for j in range(m):
                if need[i][j] > available[j]:
                    possible = False
            
            if possible:
                for j in range(m):
                    available[j] += allocation[i][j]
                
                completed.append(i)
                found = True
    
    if found == False:
        break

if len(completed) == n:
    print("\nSystem safe")
    print("Safr Sequence: ", end="")

    for i in completed:
        print(f"P{i+1} ", end=" ")
else:
    print("\nSystem in UnSAFE")
```

**Round Robin**

```python
n = int(input("Enter the number of processes: "))
processes = []

for i in range(n):
    at = int(input("Enter the arrival time(AT) for P{i+1}: "))
    bt = int(input("Enter the burst time(BT) for P{i+1}: "))
    processes.append([i+1, at, bt, bt, 0])

time_quantum = int(input("\nEnter the time quantum: "))

processes.sort(key=lambda x:x[1])

current_time = 0
completed = 0
twt = 0
ttat = 0

queue = []
index = 0

print("\nProcess\tAT\tBT\tCT\tTAT\tWT")

while completed < n:
    while index < n and processes[index][1] <= current_time:
        queue.append(processes[index])
        index += 1
    
    if len(queue) == 0:
        current_time = processes[index][1]
        continue
    
    process = queue.pop(0)

    pid, at, bt, remaining_bt, ct = process
    if remaining_bt > time_quantum:
        current_time += time_quantum
        process[3] -= time_quantum
    
    else:
        current_time += remaining_bt
        process[3] = 0

        process[4] = current_time

        ct = process[4]
        tat = ct - at
        wt = tat - bt

        twt += wt
        ttat += tat

        completed += 1

        print(f"P{pid}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")

    while index < n and process[index][1] <= current_time:
        queue.append(processes[index])
        index += 1
    
    if process[3] > 0:
        queue.append(process)

avg_wt = twt / n
avg_tat = ttat / n

print(f"\nAverage waiting time = {avg_wt:.2f}")
print(f"Average turnaround time = {avg_tat:.2f}")
```

**SJF Preemptive**

```python
n = int(input("Enter the number of processes: "))
processes = []

for i in range(n):
    at = int(input(f"Enter Arrival time (AT) for P{i+1}: "))
    bt = int(input(f"Enter Burst time (BT) for P{i+1}: "))
    processes.append([i+1, at, bt, bt, 0])

processes.sort(key=lambda x: x[1])

current_time = 0
completed = 0
twt = 0
ttat = 0

print(f"\nProcess\tAT\tBT\tCT\tTAT\tWT")

while completed < n:

    available = []

    for process in processes:
        if process[1] <= current_time and process[3] > 0:
            available.append(process)

    if len(available) == 0:
        current_time += 1
        continue

    available.sort(key=lambda x: x[3])

    process = available[0]

    pid, at, bt, remaining_bt, ct = process

    process[3] -= 1
    current_time += 1

    if process[3] == 0:

        process[4] = current_time

        ct = process[4]
        tat = ct - at
        wt = tat - bt

        twt += wt
        ttat += tat

        completed += 1

        print(f"P{pid}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")


avg_wt = twt / n
avg_tat = ttat / n

print(f"\nAverage waiting time = {avg_wt:.2f}")
print(f"Average turnaround time = {avg_tat:.2f}")
```

**SJF Non-Preemptive**

```python
n = int(input("Enter the number of processes: "))
processes = []

for i in range(n):
    at = int(input(f"Enter Arrival time (AT) for P{i+1}: "))
    bt = int(input(f"Enter Burst time (BT) for P{i+1}: "))
    processes.append([i+1, at, bt])

# First sort by Arrival Time
processes.sort(key=lambda x: x[1])

current_time = 0
twt = 0
ttat = 0

print("\nProcess\tAT\tBT\tCT\tTAT\tWT")

while len(processes) > 0:

    # If CPU is idle, take the first arrived process
    if current_time < processes[0][1]:
        current_time = processes[0][1]

    # Find all processes that have arrived
    available = []

    for process in processes:
        if process[1] <= current_time:
            available.append(process)

    # Sort available processes by Burst Time
    available.sort(key=lambda x: x[2])

    # Take the process with shortest BT
    process = available[0]

    pid, at, bt = process

    ct = current_time + bt
    tat = ct - at
    wt = tat - bt

    twt += wt
    ttat += tat

    print(f"P{pid}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")

    current_time = ct

    # Remove completed process
    processes.remove(process)

avg_wt = twt / n
avg_tat = ttat / n

print(f"\nAverage waiting time = {avg_wt:.2f}")
print(f"Average turnaround time = {avg_tat:.2f}")
```

**FCFS**

```python
n = int(input("Enter the number of processes"))
processes = []

for i in range(n):
    at = int(input(f"Enter Arrival time (AT) for P{i+1}: "))
    bt = int(input(f"Enter the busrt time (BT) for P{i+1}: "))
    processes.append([i+1,at,bt])

processes.sort(key=lambda x:x[1])
current_time = 0
twt = 0
ttat = 0

print(f"\nProcess\tAT\tBT\tCT\tTAT\tWT")

for process in processes:
    pid, at, bt = process

    if current_time < at:
        current_time = at
    
    ct = current_time + bt
    tat = ct - at
    wt = tat - bt
    twt += wt
    ttat += tat

    print(f"P{pid}\t{at}\t{bt}\t{ct}\t{tat}\t{wt}")
    current_time = ct

avg_wt = twt / n
avg_tat = ttat / n

print(f"\nAverage waiting time = {avg_wt:.2f}")
print(f"Average turnaround time = {avg_tat:.2f}")
```
